```python
# Import libraries
from ultralytics import YOLO
import torch
import os

# Check GPU availability
print(f"PyTorch version: {torch.__version__}")
print(f"CUDA available: {torch.cuda.is_available()}")
if torch.cuda.is_available():
    print(f"GPU: {torch.cuda.get_device_name(0)}")
    print(f"VRAM: {torch.cuda.get_device_properties(0).total_memory / 1e9:.2f} GB")
```

```python
import zipfile
import os

# Path to your zip file on Drive
ZIP_PATH = "/content/drive/MyDrive/plantdiseasedetection/dataset_small.zip"

# Unzip to Colab's local /content directory (faster than Drive)
print("📦 Unzipping dataset...")
with zipfile.ZipFile(ZIP_PATH, 'r') as zip_ref:
    zip_ref.extractall('/content/drive/MyDrive/')

print("✅ Unzipped!")
print("\n📁 Contents:")
!ls -la /content/dataset_small/
```

```python
# Dataset path - update this based on your setup
# For local: use relative or absolute path to dataset_small folder
# For Colab: use path after mounting Drive

# DATASET_PATH = "dataset_small/data.yaml"  # Local path
DATASET_PATH = "/content/drive/MyDrive/dataset_small/data.yaml"  # Colab path

# Verify dataset exists
if os.path.exists(DATASET_PATH):
    print(f"✅ Dataset found at: {DATASET_PATH}")
    with open(DATASET_PATH, 'r') as f:
        print("\ndata.yaml contents:")
        print(f.read()[:500] + "...")
else:
    print(f"❌ Dataset not found at: {DATASET_PATH}")
    print("Please update DATASET_PATH to point to your data.yaml file")
```

```python
# Training hyperparameters optimized for 15GB VRAM
TRAINING_CONFIG = {
    'model': 'yolov8s.pt',           # YOLOv8 Small - good balance of speed/accuracy
    'data': DATASET_PATH,
    'epochs': 100,                    # Number of training epochs
    'imgsz': 640,                     # Image size
    'batch': 32,                      # Batch size (adjust if OOM error)
    'patience': 20,                   # Early stopping patience
    'save': True,                     # Save checkpoints
    'save_period': 10,                # Save every N epochs
    'device': 0,                      # GPU device (0 for first GPU)
    'workers': 4,                     # DataLoader workers
    'project': 'plant_disease',       # Project name
    'name': 'yolov8s_rgb_ir',         # Run name
    'exist_ok': True,                 # Overwrite existing
    'pretrained': True,               # Use pretrained weights
    'optimizer': 'AdamW',             # Optimizer
    'lr0': 0.001,                     # Initial learning rate
    'lrf': 0.01,                      # Final learning rate factor
    'momentum': 0.937,                # SGD momentum
    'weight_decay': 0.0005,           # Weight decay
    'warmup_epochs': 3,               # Warmup epochs
    'warmup_momentum': 0.8,           # Warmup momentum
    'box': 7.5,                       # Box loss gain
    'cls': 0.5,                       # Class loss gain
    'dfl': 1.5,                       # DFL loss gain
    'hsv_h': 0.015,                   # HSV-Hue augmentation
    'hsv_s': 0.7,                     # HSV-Saturation augmentation
    'hsv_v': 0.4,                     # HSV-Value augmentation
    'degrees': 10,                    # Rotation augmentation
    'translate': 0.1,                 # Translation augmentation
    'scale': 0.5,                     # Scale augmentation
    'shear': 5,                       # Shear augmentation
    'flipud': 0.5,                    # Flip up-down probability
    'fliplr': 0.5,                    # Flip left-right probability
    'mosaic': 1.0,                    # Mosaic augmentation
    'mixup': 0.1,                     # Mixup augmentation
    'copy_paste': 0.1,                # Copy-paste augmentation
}

print("📋 Training Configuration:")
for key, value in TRAINING_CONFIG.items():
    print(f"  {key}: {value}")
```

```python
# Load YOLOv8s model
model = YOLO(TRAINING_CONFIG['model'])
print(f"✅ Model loaded: {TRAINING_CONFIG['model']}")
# Start training
print("🚀 Starting training...")
print("="*60)

results = model.train(**TRAINING_CONFIG)

print("="*60)
print("✅ Training complete!")
```

```python
# Validate on test set
print("\n📊 Evaluating on test set...")
metrics = model.val(data=DATASET_PATH, split='test')

print(f"\n📈 Test Results:")
print(f"  mAP50: {metrics.box.map50:.4f}")
print(f"  mAP50-95: {metrics.box.map:.4f}")
print(f"  Precision: {metrics.box.mp:.4f}")
print(f"  Recall: {metrics.box.mr:.4f}")
```

```python
# Export best model to different formats
best_model_path = "/content/runs/detect/plant_disease/yolov8s_rgb_ir/weights/best.pt"

# Load best model
best_model = YOLO(best_model_path)

# Export to ONNX (for deployment)
print("\n📦 Exporting model...")
best_model.export(format='onnx')
print("✅ Model exported to ONNX format!")
```
